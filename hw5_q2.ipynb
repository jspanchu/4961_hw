```python
from sklearn.datasets import load_breast_cancer

data = load_breast_cancer()

print(data.feature_names)
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import classification_report, accuracy_score

# Convert the dataset to a pandas DataFrame for easier feature selection
df = pd.DataFrame(data.data, columns=data.feature_names)
y = data.target

# Select only the specified features
selected_features = ['worst compactness', 'worst concavity', 'worst area']
X = df[selected_features]

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# Scale the features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

# 2.a) What is the accuracy of the classifier with linear kernel

```python
# Initialize and train the SVM classifier
svm_linear_classifier = SVC(kernel='linear', random_state=42)
svm_linear_classifier.fit(X_train_scaled, y_train)

# Make predictions and evaluate
y_pred = svm_linear_classifier.predict(X_test_scaled)
accuracy = accuracy_score(y_test, y_pred)

print(f"Selected Features: {selected_features}")
print(f"Test Accuracy: {accuracy:.4f}\n")
```

# 2.b) What is the accuracy of the classifier with rbf kernel and C=2?

```python
# Initialize and train the SVM classifier
svm_rbf_classifier = SVC(kernel='rbf', C=2.0, random_state=42)
svm_rbf_classifier.fit(X_train_scaled, y_train)

# Make predictions and evaluate
y_pred = svm_rbf_classifier.predict(X_test_scaled)
accuracy = accuracy_score(y_test, y_pred)

print(f"Selected Features: {selected_features}")
print(f"Test Accuracy: {accuracy:.4f}\n")
```
